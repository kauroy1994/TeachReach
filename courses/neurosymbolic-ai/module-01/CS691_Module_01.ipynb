## Give each claim a name

A predicate names a property or relationship and tells us which positions to fill. Available(item) takes one item. Requested(person, item) takes a person and an item, in that order. Filling all the positions gives a statement we can ask about or record as a fact.

Requested(Ava, C1) says who requested which item. In Python we can keep the predicate name and its values together in a tuple, an ordered group in parentheses. Quotation marks mark names as text.

**Predict before running the cell.** The note says “C1 was requested by Ava.” Which position should C1 occupy in Requested?

````python
fact = ("Requested", "Ava", "C1")
predicate, person, item = fact
print(person, predicate.lower(), item)
````

<details><summary>Reveal output and explanation</summary>

```text
Ava requested C1
```

C1 is the second argument, the item. The wording of the note changed, but Ava still fills the person position.

</details>

## Ask a question with a blank

A record can answer more than a yes-or-no question. In Requested(Ava, x), x is a variable, a placeholder for an item. Look for records whose person is Ava and read the item from each match.

With two requests by Ava and one by Bo, “What did Ava request?” has two answers. Each answer names an item that makes the statement match a record. The Python loop visits each person–item pair.

**Predict before running the cell.** How would you answer “Who requested C1?” with these same records?

````python
requests = [("Ava", "C1"), ("Bo", "T2"), ("Ava", "M1")]
for person, item in requests:
    if person == "Ava":
        print(item)
````

<details><summary>Reveal output and explanation</summary>

```text
C1
M1
```

Fix the item as C1 and read the person from each match. Only the first row has that item, so the answer in this record is Ava.

</details>

## Leave room for an unknown

A desk record can be incomplete. In an open-world reading of the record, a missing fact leaves the corresponding claim unknown. A recorded negative claim can provide evidence against it. Python’s membership test answers a narrower question about what is in a collection.

C1 has an available label, and T2 has an unavailable label. Nothing in the record settles M1’s status. The Python dictionary pairs each recorded item with its status. Its get method supplies a default for a missing key.

**Predict before running the cell.** A note says that Bo requested M1. What does that change about M1’s availability?

````python
status = {"C1": "available", "T2": "unavailable"}
for item in ["C1", "T2", "M1"]:
    print(item, status.get(item, "unknown"))
````

<details><summary>Reveal output and explanation</summary>

```text
C1 available
T2 unavailable
M1 unknown
```

It adds a request. Availability is a separate claim, so M1’s availability remains unknown until there is evidence about that claim.

</details>
