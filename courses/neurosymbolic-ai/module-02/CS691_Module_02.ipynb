## Say who a statement covers

First-order logic lets us make statements about individual things and relationships among them. The domain is the collection of individuals under discussion. A quantifier tells us whether the claim must hold for every individual in that domain or for at least one. “Every borrower on this list completed camera orientation” asks about each listed person. The list fixes the domain for this check.

For this example the list is complete and each person has an explicit yes or no for orientation. Ava has yes and Bo has no. Python’s all returns True only if every value it checks is true.

**Predict before running the cell.** Which single record is enough to disprove “every listed borrower completed orientation”?

````python
completed = {"Ava": True, "Bo": False}
print(all(completed.values()))
````

<details><summary>Reveal output and explanation</summary>

```text
False
```

Bo’s explicit no is a counterexample. Ava’s yes establishes the claim for Ava only.

</details>

## Find a witness

“At least one borrower on the list completed orientation” needs one matching person. The symbol ∃ means “there exists.” The symbol ∀ means “for every.” The same record can make the first claim true and the second one false.

Ava is a witness, a specific individual who satisfies the condition. Python’s items() supplies a name and its recorded value together. The loop prints each name with a yes.

**Predict before running the cell.** If Bo also completes orientation, is Ava still a witness? How many witnesses are there?

````python
completed = {"Ava": True, "Bo": False}
for person, yes in completed.items():
    if yes:
        print(person)
````

<details><summary>Reveal output and explanation</summary>

```text
Ava
```

Yes. The claim allows any positive number of witnesses. With both records set to yes, Ava and Bo each provide one.

</details>

## Apply an if–then policy

A rule gives a conclusion that follows when its conditions hold. At our fictional desk, anyone who completes camera orientation meets the eligibility requirement for camera loans. Write ∀p (Oriented(p) → Eligible(p)). Read the arrow as “if the statement on the left holds, the statement on the right follows.”

Starting from one recorded orientation, substitute Ava for p on both sides. Oriented(Ava) gives Eligible(Ava). The set below represents the names with a confirmed orientation. The loop applies the same rule to every name in it.

**Predict before running the cell.** Someone tells you Bo is eligible. Does the arrow also establish that Bo completed orientation?

````python
oriented = {"Ava"}
eligible = set()
for person in oriented:
    eligible.add(person)
print(sorted(eligible))
````

<details><summary>Reveal output and explanation</summary>

```text
['Ava']
```

No. This rule gives a sufficient route to eligibility. Another policy might grant eligibility by a different route. Reasoning backwards from Eligible(Bo) would require more information.

</details>

## Keep names aligned across conditions

For the camera requests in this example, the desk marks a request ready for review when the requester is eligible and the requested item is available. The conditions are joined by “and.” The same person and item must keep their identities throughout one application of the rule.

Requested(p, i) and Eligible(p) and Available(i) together give Ready(p, i). For Ava’s C1 request, check Ava’s eligibility and C1’s availability. A different person’s eligibility cannot fill Ava’s position.

**Predict before running the cell.** C2 is also available. Can the rule mark Ava’s C2 request ready using just these records?

````python
requests = [("Ava", "C1"), ("Bo", "C1")]
eligible = {"Ava"}
available = {"C1"}
for person, item in requests:
    if person in eligible and item in available:
        print(person, item)
````

<details><summary>Reveal output and explanation</summary>

```text
Ava C1
```

No. The request list ties Ava to C1. Available(C2) cannot supply the missing Requested(Ava, C2).

</details>

## Follow the person and camera through the rule

Each pass through the loop picks up one person and the item that person requested. The word in checks whether a value belongs to a set. The word and requires both checks to pass on that same row before Python prints its two names.

Replace this line:
eligible = {"Ava"}
with:
eligible = {"Bo"}
What will print?

````python
requests = [("Ava", "C1"), ("Bo", "C1")]
eligible = {"Ava"}
available = {"C1"}
for person, item in requests:
    if person in eligible and item in available:
        print(person, item)
````

<details><summary>Worked answer after your change</summary>

Bo C1

Both people requested C1, but the revised eligible set contains only Bo. Camera C1 still passes the availability check.

</details>
