## Read a fact as an edge

A graph has nodes for things and edges for relationships. In our knowledge graph, each edge carries a label with a defined meaning. A triple lists a starting node (subject), the relationship (predicate), and the ending node (object).

The triple (Ava, requested, C1) draws a directed edge from Ava to C1. The general logical notation from Module 01 was Requested(Ava, C1). These are two ways to record the same relationship.

**Predict before running the cell.** Which triple would record that C1 is stored in Cabinet7?

````python
triple = ("Ava", "requested", "C1")
subject, relation, obj = triple
print(subject, "—", relation, "→", obj)
````

<details><summary>Reveal output and explanation</summary>

```text
Ava — requested → C1
```

Use (C1, storedIn, Cabinet7). The edge starts at the item and ends at the place where it is stored.

</details>

## Distinguish an item from a class

An individual is one particular thing, such as the camera numbered C1. A class is a category, such as Camera. A type edge connects an individual to a class it belongs to. Several individuals can have the same type.

C1 and C2 can both be cameras while occupying different shelves. A query for cameras matches each item whose type edge points to Camera.

**Predict before running the cell.** The desk moves C1 to a different cabinet. What type edge should change?

````python
types = [("C1", "Camera"), ("C2", "Camera"), ("T2", "Tripod")]
for item, category in types:
    if category == "Camera":
        print(item)
````

<details><summary>Reveal output and explanation</summary>

```text
C1
C2
```

None for that move. C1 is still a Camera. Its location edge is the one affected.

</details>

## Let a class carry meaning

An ontology specifies a vocabulary of classes and relationships and statements about their meanings. Our small inventory ontology says that Camera is a subclass of Equipment. That says every member of Camera is also a member of Equipment.

Combine C1 has type Camera with Camera is a subclass of Equipment. You can now infer that C1 has type Equipment. This code takes one step through a hierarchy with one parent per class in the example.

**Predict before running the cell.** If all you know is that item E7 is Equipment, can this hierarchy tell you it is a Camera?

````python
item, category = "C1", "Camera"
parent = {"Camera": "Equipment", "Tripod": "Equipment"}
print(item, "has type", parent[category])
````

<details><summary>Reveal output and explanation</summary>

```text
C1 has type Equipment
```

No. Camera membership implies Equipment membership. Equipment can also contain tripods or other kinds of item.

</details>

## Follow a path with a shared node

To find where staff store a requested item, follow the requested edge from the person to the item. Keep that item as the starting point of the storedIn edge. Two facts answer a question that neither fact answers alone.

Ava requested C1. C1 is stored in Cabinet7 and C2 is stored elsewhere. Matching the intermediate item keeps C2’s storage location out of the answer about C1.

**Predict before running the cell.** Suppose the location for C1 is missing but Cabinet2 is recorded for C2. What can the path answer?

````python
requests = [("Ava", "C1")]
location = {"C1": "Cabinet7", "C2": "Cabinet2"}
for person, item in requests:
    if item in location:
        print(person, item, location[item])
````

<details><summary>Reveal output and explanation</summary>

```text
Ava C1 Cabinet7
```

It still identifies Ava’s requested item as C1. It supplies no location for that item. The edge from C2 cannot complete a path through C1.

</details>
