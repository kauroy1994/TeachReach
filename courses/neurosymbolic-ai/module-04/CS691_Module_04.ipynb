## Make one inference

Inference means obtaining a conclusion from premises using a rule. A premise is a statement a step depends on. For this fictional camera desk, rule R1 says that an oriented borrower is eligible. A recorded fact can supply R1’s premise.

The orientation register confirms Oriented(Ava). Apply R1 to Ava, keeping her name fixed. We can add Eligible(Ava) to our working set of facts and retain the original record.

**Predict before running the cell.** If R1 is applied twice to the same record, how many distinct eligibility facts about Ava exist?

````python
facts = {("Oriented", "Ava")}
if ("Oriented", "Ava") in facts:
    facts.add(("Eligible", "Ava"))
print(sorted(facts))
````

<details><summary>Reveal output and explanation</summary>

```text
[('Eligible', 'Ava'), ('Oriented', 'Ava')]
```

One. Both applications yield the same claim, Eligible(Ava). A set keeps one copy.

</details>

## Use one conclusion in the next rule

Forward chaining starts with available facts and adds the conclusions of rules whose conditions hold. A new conclusion can help another rule apply. Our second rule, R2, combines a request, the borrower’s eligibility, and the item’s availability to mark that request ready for review.

For this cell, the request is Ava’s request for C1. We shorten the four conditions to words because the person and item stay fixed here. Python’s issubset asks whether every required condition is present.

**Predict before running the cell.** Suppose R2 is checked before R1 and just once. Which conclusion would the program miss?

````python
facts = {"requested", "oriented", "available"}
if "oriented" in facts:
    facts.add("eligible")
if {"requested", "eligible", "available"}.issubset(facts):
    facts.add("ready")
print("ready" in facts)
````

<details><summary>Reveal output and explanation</summary>

```text
True
```

It would miss ready because eligible has not yet been added. A general forward-chaining procedure revisits affected rules until no new facts follow, assuming a finite set of possible facts in this example.

</details>

## Save the reasons along with the answer

A proof trace records a conclusion, the rule that produced it, and the supporting premises. To audit Ready(Ava, C1), work backward from that conclusion. Each branch should reach a recorded premise or another justified step.

Here each short name refers to Ava and C1 as on the previous slide. The dictionary stores the rule and the exact premises for each derived claim. The trace for ready points to eligible, whose own trace leads to the orientation record.

**Predict before running the cell.** Which original records must a clerk inspect at the leaves of this trace?

````python
trace = {"ready": ("R2", ["requested", "eligible", "available"]),
         "eligible": ("R1", ["oriented"])}
for claim in ["ready", "eligible"]:
    rule, premises = trace[claim]
    print(claim, rule, " + ".join(premises))
````

<details><summary>Reveal output and explanation</summary>

```text
ready R2 requested + eligible + available
eligible R1 oriented
```

The request by Ava for C1, Ava’s orientation record, and the availability record for C1. Eligible is an intermediate conclusion justified by R1.

</details>

## Find the first missing premise

A clerk asks why Ava’s request did not reach ready for review. Check the conditions for R2. Suppose the desk can establish the request and eligibility, but has no availability record for C1. A useful explanation names the missing evidence.

The minus sign between two Python sets takes the set difference. Here it finds the requirements that have no support in the current records. The output tells the clerk which record to look for.

**Predict before running the cell.** Return to the complete proof on the preceding slide. If its C1 availability record was out of date, which conclusion needs checking?

````python
required = {"requested", "eligible", "available"}
supported = {"requested", "eligible"}
missing = required - supported
print(sorted(missing))
````

<details><summary>Reveal output and explanation</summary>

```text
['available']
```

Inspect the availability premise and recompute any conclusion that used it. The rule application could have been valid for the recorded inputs even though one input was inaccurate.

</details>
