## Apply the rule and cite the whole reason

An evidence-based decision needs both the facts and the rule connecting them to the conclusion. R4 states that Ava’s reservation for C1 is active. T7 states that Ava completed camera orientation. P1 explains why those two facts together make Ava eligible.

A reader can challenge any of the three pieces. A missing fact and a missing rule leave different gaps in the explanation.

## Ask for the particular evidence you lack

Abstaining means withholding a conclusion the evidence cannot establish. Make that response actionable by naming the unresolved requirement. A missing orientation record calls for a lookup. An explicit current record that orientation is incomplete establishes a failed requirement under P1. These situations warrant different answers.

Here R4 already confirms an active reservation. The three cases vary only what you learn about orientation. Python’s None marks missing information.

**Predict before running the cell.** What should you do if two equally current orientation records disagree and the desk has no rule for choosing one?

````python
for orientation in [None, "incomplete", "complete"]:
    if orientation is None:
        print("Need a current camera orientation record.")
    elif orientation == "incomplete":
        print("Cannot issue under P1: orientation incomplete.")
    else:
        print("Eligible under P1.")
````

<details><summary>Reveal output and explanation</summary>

```text
Need a current camera orientation record.
Cannot issue under P1: orientation incomplete.
Eligible under P1.
```

Report the conflict, identify both records and request a corrected or authoritative status. Choosing the first record returned would make the decision depend on search order.

</details>

## Check the proposed answer end to end

Combine the pieces in sequence. Retrieve the two records, apply P1, compare the derived verdict to the supplied language-model draft, and publish a conclusion with its evidence. A fluent draft can still guess the verdict correctly by chance. Its citations also have to cover the evidence that the rule requires.

The sample draft proposes “eligible” and cites P1, R4 and D8. The verdict matches the rule, but D8 concerns the desk location. Run the check to see which source is missing. Then replace D8 with the ID it requests and run it again.

**Predict before running the cell.** Does adding T7 automatically make every sentence in a longer draft safe to publish?

````python
facts = {("Ava", "C1"): ("active", "R4"),
         ("Ava", "camera orientation"): ("complete", "T7")}
reservation, rid = facts[("Ava", "C1")]
training, tid = facts[("Ava", "camera orientation")]
required_ids = {"P1", rid, tid}
eligible = reservation == "active" and training == "complete"
draft = {"eligible": True, "sources": {"P1", "R4", "D8"}}
if draft["eligible"] != eligible:
    print("Revise the verdict.")
elif not required_ids.issubset(draft["sources"]):
    print("Missing source:", ", ".join(sorted(required_ids - draft["sources"])))
else:
    print("Verdict and required sources match.")
````

```text
Missing source: T7
```

No. It supplies the missing source for orientation. A new assertion about pickup time, fees or any other claim would need its own check. The code checks only the specified verdict and required IDs.

</details>
