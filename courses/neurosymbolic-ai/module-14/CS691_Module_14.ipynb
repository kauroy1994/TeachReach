## Why was the extra clerk there?

The supervisor assigns extra help when a shift is busy. Crowding can also increase wait time. A factor that influences both the staffing decision and the wait is a possible confounder. The two observed shifts give us no equally busy pair with different staffing.

Use 0 for quiet or no extra clerk and 1 for busy or an extra clerk. Each row records busy, extra and minutes waited. Grouping by busy shows which choices were actually observed together.

**Predict before running the cell.** Can we find out what happened on a busy shift without the extra clerk by filtering this table?

````python
shifts = [(0, 0, 4), (1, 1, 8)]
for busy in (0, 1):
    choices = [extra for load, extra, wait in shifts if load == busy]
    print(busy, choices)
````

<details><summary>Reveal output and explanation</summary>

```text
1 [1]
```

No row contains that combination. Filtering can select observations the desk has, but cannot supply the missing comparison.

</details>

## Change one decision inside a stated model

An intervention sets a choice deliberately. To practice, assume the wait starts at 4 minutes, a busy desk adds 6 and an extra clerk saves 2, with everything else fixed. These are assumptions of a toy model. They let us calculate the outcome of each staffing choice on the same busy shift.

Set busy to 1 in both calculations. Compare a clerk count setting of 1 (extra help) with 0 (usual staffing). The unobserved alternative for that shift is its counterfactual outcome under this model.

**Predict before running the cell.** Why does this difference have the opposite sign from the observed four-minute difference?

````python
def wait(busy, extra):
    return 4 + 6 * busy - 2 * extra

without = wait(1, 0)
with_extra = wait(1, 1)
print(without, with_extra, with_extra - without)
````

<details><summary>Reveal output and explanation</summary>

```text
10 8 -2
```

The observed comparison changed crowding and staffing together. Here the same busy shift is considered under two staffing choices. The model assigns a two-minute reduction to the extra clerk, giving 8 minus 10, or −2.

</details>

## Two explanations can fit the same log

To see what the log leaves open, try another model. Suppose a busy shift adds 4 minutes and the extra clerk changes nothing. Both models reproduce the two observed waits. Their predictions disagree for a busy shift with usual staffing.

Model A is the previous rule. Model B ignores staffing. Compare predictions on the observed rows and then on the combination the log lacks.

**Predict before running the cell.** Which additional staffing comparison would help distinguish these explanations?

````python
def model_a(busy, extra):
    return 4 + 6 * busy - 2 * extra

def model_b(busy, extra):
    return 4 + 4 * busy

for busy, extra in [(0, 0), (1, 1), (1, 0)]:
    print(model_a(busy, extra), model_b(busy, extra))
````

<details><summary>Reveal output and explanation</summary>

```text
4 4
8 8
10 8
```

Compare otherwise comparable busy shifts with and without an extra clerk. A study that randomly assigns extra help across many such shifts could reduce systematic differences between staffing groups.

</details>
